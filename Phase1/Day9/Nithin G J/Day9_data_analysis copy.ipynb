```python
import pandas as pd
import numpy as np
# create sample datasets
data={
    "Name":["Naveen","Akash","Devika","Shreya","Nithin","Komal","Bhuvan","Kaveri","Vasthsal","Disha"],
    "City":["Bengaluru","Mysuru","Bengaluru","Hubli","Bengaluru","Mysuru","Hubli","Bengaluru","Mysuru","Bengaluru"],
    "Marks":[85,72,91,55,78,65,48,88,70,92],
    "Age":[20,21,19,22,20,21,20,19,21,20]
    
}

df=pd.DataFrame(data)
df
```

```python
#groupBy city-average marks per city
city_avg=df.groupby("City")["Marks"].mean()
print("Average Marks per city:")
print(city_avg)
```

```python
#Multiple groupby operations
print("Sum of marks per city:")
print(df.groupby("City")["Marks"].sum())

print("\nCount of students per city:")
print(df.groupby("City")["Marks"].count())

print("\nMaximum marks per city:")
print(df.groupby("City")["Marks"].max())


print("\nMinimum marks per city:")
print(df.groupby("City")["Marks"].min())

```

```python
#Multiple Aggregations at once using agg()
city_summary=df.groupby("City")["Marks"].agg(["mean","max","min","count"])
city_summary.columns=["Average","Highest","Lowest","Studentcount"]
print("city wise summary:")
print(city_summary)
```

```python
city_average_age=df.groupby("City")["Age"].mean()
print(city_average_age)
```

```python
city_average=df.groupby("City")["Age"].agg(["mean","max","min"])
city_average.columns=["Average_age","Highest_age","Lowest_age"]
print("city average age summary:")
print(city_average)
```

```python
#describe()-full explanation
print("Full describe():")
print(df.describe())
```

```python
#Individual statistics
print("Mean marks:",df["Marks"].mean())
print("Median marks:",df["Marks"].median())
print("Std Deviations:",df["Marks"].std())
print("variance:",df["Marks"].var())
print("25th percentile:",df["Marks"].quantile(0.25))
print("75th percentile:",df["Marks"].quantile(0.75))
print("correlation between age and marks:")
print(df[["Age","Marks"]].corr())

```

```python
#value Counts
print("Student per City:")
print(df["City"].value_counts())

print("\nPercentage per city:")
print(df["City"].value_counts(normalize=True)*100)
```

```python
print("Full describe():")
print(df["Marks"].describe())
```

```python
print("Median marks:",df["Marks"].median())
print("Student per City:")
vrv=df["City"].value_counts()
print(vrv)
print(vrv.head(1))
```

```python
#data explration
#Top 3 Students by Marks
top3=df.nlargest(3,"Marks")
print("top 3 students:")
print(top3[["Name","Marks","City"]])

#Botttom 3 students by Marks
bottom3=df.nsmallest(3,"Marks")
print("Bottom 3 students:")
print(bottom3[["Name","Marks","City"]])
```

```python
#conditional analysis
#above average students
avg_marks=df["Marks"].mean()
above_avg=df[df["Marks"]>avg_marks]
below_avg=df[df["Marks"]<avg_marks]

print(f"Class average:{avg_marks:.2f}")
print(f"\nStudents above average({len(above_avg)}):")
print(above_avg[["Name","Marks"]])

print(f"\nStudents below average({len(below_avg)}):")
print(below_avg[["Name","Marks"]])
```

```python
#Trend:city-wise performance
city_performance=df.groupby("City")["Marks"].mean().sort_values(ascending=False)
print("City Performance ranking (Average marks):")
print(city_performance)

#Trend:Pass rate per city
df["Result"]=df["Marks"].apply(lambda x:"Pass" if x>=60 else "Fail")
pass_rate=df.groupby("City")["Result"].apply(lambda x:(x=="Pass").sum()/len(x)*100)
print("\nPass Rate Per City(%)")
print(pass_rate.round(2))
```

```python
top3=df.nlargest(2,"Marks")
print("top 3 students:")
print(top3[["Name","Marks","City"]])
bottom3=df.nsmallest(2,"Marks")
print("Bottom 3 students:")
print(bottom3[["Name","Marks","City"]])
```

```python
avg_marks=df["Marks"].mean()
above_avg=df[df["Marks"]>avg_marks]
below_avg=df[df["Marks"]<avg_marks]
print(f"average marks:",avg_marks)
print(f"Students above average({len(above_avg)})")
print(f"\nStudents below average({len(below_avg)})")
```

```python
print("="*50)
print("     Student Summary Report")
print("="*50)

print(f"\nTotal Students: {len(df)}")
print(f"Average Marks: {df["Marks"].mean():.2f}")
print(f"Highest Marks: {df['Marks'].max()}")
print(f"Lowest Marks: {df['Marks'].min()}")
print(f"Median Marks: {df['Marks'].median()}")

print("\n--- Pass/Fail Summary ---")
result_counts = df["Result"].value_counts()
for result, count in result_counts.items():
    print(f"{result}: {count} students")

print("\n--- Grade Summary ---")
df["Grade"] = df["Marks"].apply(lambda x: "A" if x >= 80 else "B" if x >= 60 else "C")
grade_counts = df["Grade"].value_counts().sort_index()
for grade, count in grade_counts.items():
    print(f"Grade {grade}: {count} students")

print("\n--- City-wise Average Marks ---")
city_avg = df.groupby("City")["Marks"].mean().sort_values(ascending=False)
for city, avg in city_avg.items():
    print(f"{city}: {avg:.2f}")

print("\n--- Top 3 Students ---")
top3 = df.nlargest(3, "Marks")
for i, row in top3.iterrows():
    print(f"{row['Name']} - {row['Marks']} marks ({row['City']})")

print("\n--- Students Needing Attention (Marks < 60) ---")
weak = df[df["Marks"] < 60]
if len(weak) == 0:
    print("All Students Passed.")
else:
    for i, row in weak.iterrows():
        print(f"{row['Name']} - {row['Marks']} marks")

print("="*50)
print("   END OF REPORT")
print("="*50)
```

```python
#Save summary to csv
summary_data={
    "Metric":[
        "Total Students",
        "Average Marks",
        "Highest Marks",
        "Lowest Marks",
        "Median Marks",
        "Pass Count",
        "Fail Count"
    ],
    "Value":[
        len(df),
        round(df["Marks"].mean(),2),
        df["Marks"].max(),
        df["Marks"].min(),
        df["Marks"].median(),
        len(df[df["Result"]=="pass"]),
        len(df[df["Result"]=="Fail"])
    ]
}

summary_df=pd.DataFrame(summary_data)
summary_df.to_csv("students_summary_report.csv",index=False)
print("Sumarry report saved to student_summary_report.csv ")
summary_df
```

```python
sales_data = {

    "Product": ["Pen","Notebook","Pen","Pencil","Notebook",

                "Pen","Pencil","Notebook","Pen","Pencil"],

    "City": ["Bengaluru","Mysuru","Hubli","Bengaluru","Bengaluru",

             "Mysuru","Hubli","Bengaluru","Hubli","Mysuru"],

    "Quantity": [10, 3, 7, 15, 5, 8, 12, 4, 6, 9],

    "Price": [20, 50, 20, 5, 50, 20, 5, 50, 20, 5]

}
df=pd.DataFrame(sales_data)
df["TotalAmount"]=df["Quantity"]*df["Price"]
print(df["TotalAmount"])


print("total  amount:")
print(df.groupby("Product")["Price"].sum())

print("\ntotal Quantity:")
print(df.groupby("Product")["Price"].count())

print("\naverage price")
print(df.groupby("Product")["Price"].mean())

print("\n using agg()")
vrv=df.groupby("Product")["Price"].agg(["mean","max","min","count"])
vrv.columns=["mean","Highest price","Lowest price","total products"]
print(vrv)
print("total  amount by city :")
print(df.groupby("City")["Price"].sum())

print("\ntotal transactions:")
print(df.groupby("City")["Price"].count())

print("\n using agg()")
vrv1=df.groupby("City")["Price"].agg(["mean","max","min","count"])
vrv1.columns=["mean","Highest price","Lowest price","total transactions"]
print(vrv1)
```

```python
print(df["TotalAmount"].describe())
print("Mean:", df["TotalAmount"].mean())
print("Median:", df["TotalAmount"].median())
print("Std Dev:", df["TotalAmount"].std())
print(df.nlargest(3, "TotalAmount"))
print(df.nsmallest(3, "TotalAmount"))
print("Highest sales product:",df.groupby("Product")["TotalAmount"].sum().idxmax())
print("Highest sales city:",df.groupby("City")["TotalAmount"].sum().idxmax())
```

```python
revenue_product = df.groupby("Product")["TotalAmount"].sum()
revenue_city = df.groupby("City")["TotalAmount"].sum()

summary_data = {
    "Metric": [
        "Total Transactions",
        "Total Revenue",
        "Average Transaction Amount",
        "Best Selling Product",
        "Best Performing City"
    ],
    "Value": [
        len(df),
        df["TotalAmount"].sum(),
        round(df["TotalAmount"].mean(), 2),
        revenue_product.idxmax(),
        revenue_city.idxmax()
    ]
}

summary_df = pd.DataFrame(summary_data)

summary_df.to_csv("sales_summary_report.csv", index=False)

print("Summary report saved to sales_summary_report.csv")
summary_df
```
