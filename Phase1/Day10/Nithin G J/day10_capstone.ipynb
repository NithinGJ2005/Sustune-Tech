```python
# Part 0 – Data Creation

import pandas as pd
import numpy as np

data = {
    "Name": ["Navami", "Akash", "Devika", "Shreya", "Nithin",
             "Komal", "Bhuvan", "Kaveri", "Vathsala", "Disha",
             "Vasukeesha", "Akhil", "Rajeshwari", "Mandara", "Thrishanth"],
    "Age": [20, 21, 19, 22, 20, 21, 20, 19, 21, 20, 22, 21, 20, 19, 21],
    "City": ["Bengaluru", "Mysuru", "Bengaluru", "Hubli", "Bengaluru",
             "Mysuru", "Hubli", "Bengaluru", "Mysuru", "Bengaluru",
             "Hubli", "Bengaluru", "Mysuru", "Bengaluru", "Hubli"],
    "Marks": [85, 72, 91, 55, 78, 65, 48, 88, 70, 92,
              58, 83, 67, 45, 76],
    "Subject": ["Python", "Python", "Data Analysis", "Python", "Data Analysis",
                "Python", "Data Analysis", "Python", "Data Analysis", "Python",
                "Python", "Data Analysis", "Python", "Data Analysis", "Python"]
}

df = pd.DataFrame(data)
df
```

```python
# Part 1 – Data Loading and Validation

# Save to CSV
df.to_csv("students_capstone.csv", index=False)

# Load with error handling
try:
    df = pd.read_csv("students_capstone.csv")
    print("File loaded successfully")
except FileNotFoundError:
    print("File not found")

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Basic info
print("\nShape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nData Types:")
print(df.dtypes)
```

```python
# Part 2 – Data Processing

# Result column
df["Result"] = np.where(df["Marks"] >= 60, "Pass", "Fail")

# Grade column
def assign_grade(marks):
    if marks >= 80:
        return "A"
    elif marks >= 60:
        return "B"
    else:
        return "C"

df["Grade"] = df["Marks"].apply(assign_grade)

# Passed students
print("\nPassed Students:")
print(df[df["Result"] == "Pass"])

# Grade A students
print("\nGrade A Students:")
print(df[df["Grade"] == "A"])

# Top 5 by marks
print("\nTop 5 Students:")
print(df.sort_values(by="Marks", ascending=False).head(5))
```

```python
# Part 3 – NumPy Analysis

marks_array = df["Marks"].to_numpy()

print("Sum:", marks_array.sum())
print("Mean:", marks_array.mean())
print("Max:", marks_array.max())
print("Min:", marks_array.min())
print("Std Dev:", marks_array.std())

# Scores above average
print("\nAbove Average Scores:")
print(marks_array[marks_array > marks_array.mean()])

# Sorted descending
print("\nSorted Marks (Desc):")
print(np.sort(marks_array)[::-1])

# 2D array: Python vs Data Analysis
python_marks = df[df["Subject"] == "Python"]["Marks"].to_numpy()
da_marks = df[df["Subject"] == "Data Analysis"]["Marks"].to_numpy()

min_len = min(len(python_marks), len(da_marks))
marks_2d = np.vstack((python_marks[:min_len], da_marks[:min_len]))

print("\n2D Array:")
print(marks_2d)

print("\nRow Means:")
print(marks_2d.mean(axis=1))
```

```python
# Part 4 – Pandas Analysis

# City-wise aggregation
print("\nCity-wise Statistics:")
print(df.groupby("City")["Marks"].agg(["mean", "max", "min", "count"]))

# Grade-wise count
print("\nGrade-wise Count:")
print(df.groupby("Grade").size())

# Subject-wise average
print("\nSubject-wise Average Marks:")
print(df.groupby("Subject")["Marks"].mean())

# Describe
print("\nMarks Description:")
print(df["Marks"].describe())

# Top & bottom students
print("\nTop 3 Students:")
print(df.nlargest(3, "Marks")[["Name", "Marks"]])

print("\nBottom 3 Students:")
print(df.nsmallest(3, "Marks")[["Name", "Marks"]])

# Pass rate per city
pass_rate = df[df["Result"] == "Pass"].groupby("City").size() / df.groupby("City").size() * 100
print("\nPass Rate per City (%):")
print(pass_rate)
```

```python
# Part 5 – Summary Report

total_students = len(df)
avg_marks = df["Marks"].mean()
max_marks = df["Marks"].max()
min_marks = df["Marks"].min()
median_marks = df["Marks"].median()
pass_count = (df["Result"] == "Pass").sum()
fail_count = (df["Result"] == "Fail").sum()
pass_rate = (pass_count / total_students) * 100

grade_summary = df["Grade"].value_counts()
city_avg = df.groupby("City")["Marks"].mean()

top_3 = df.nlargest(3, "Marks")
attention = df[df["Marks"] < 60]

print("""
PHASE 1 CAPSTONE - STUDENT ANALYSIS REPORT
==================================================
""")
print(f"Total Students     : {total_students}")
print(f"Average Marks      : {avg_marks:.2f}")
print(f"Highest Marks      : {max_marks}")
print(f"Lowest Marks       : {min_marks}")
print(f"Median Marks       : {median_marks}")
print(f"Pass Count         : {pass_count}")
print(f"Fail Count         : {fail_count}")
print(f"Pass Rate          : {pass_rate:.2f}%")

print("\n--------------------------------------------------")
print("GRADE SUMMARY")
for grade, count in grade_summary.items():
    print(f"Grade {grade}          : {count} students")

print("\n--------------------------------------------------")
print("CITY-WISE AVERAGE MARKS")
for city, avg in city_avg.items():
    print(f"{city:<16}: {avg:.2f}")

print("\n--------------------------------------------------")
print("TOP 3 STUDENTS")
for i, row in top_3.iterrows():
    print(f"{row['Name']} - {row['Marks']}")

print("\n--------------------------------------------------")
print("STUDENTS NEEDING ATTENTION (Marks < 60)")
for i, row in attention.iterrows():
    print(f"{row['Name']} - {row['Marks']}")

print("==================================================")

# Save summary CSV
summary_df = df[["Name", "Marks", "Grade", "Result", "City"]]
summary_df.to_csv("capstone_summary_report.csv", index=False)
```
