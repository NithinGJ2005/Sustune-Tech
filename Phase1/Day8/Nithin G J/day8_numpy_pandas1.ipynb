```python
import numpy as np
#creating a numpy array
numbers=np.array([3,10,7,2,15,8,21,14])

print("Array:",numbers)
print("Type:",type(numbers))
print("data type",numbers.dtype)
print("Shape:",numbers.shape)
print("Length",len(numbers))
```

```python
#array operations
numbers=np.array([3,10,7,2,15,8,21,14])
print("Original",numbers)
print("ADD 10:",numbers+10)
print("Multiply by 2",numbers*2)
print("Square",numbers**2)
print("Greater than 10:",numbers>10)
print("Sum:",np.sum(numbers))
print("Mean:",np.mean(numbers))
print("Max:",np.max(numbers))
print("Min:",np.min(numbers))
print("Sorted:",np.sort(numbers))
print("Greater than 10:",numbers[numbers>10])
print("descending Sorted:",np.sort(numbers)[::-1])
```

```python
#2D array
marks_2d=np.array([
    [85,72,91],
    [65,78,88],
    [90,55,70]
])
print("2D array:\n",marks_2d)
print("Shape",marks_2d.shape)
print("First row:",marks_2d[0])
print("First Column:",marks_2d[:,0])
print("Element at row 1 ,col 2:",marks_2d[0][1])
print("Element at row 1 ,col 2:",marks_2d[0,1])
print("Column means:",np.mean(marks_2d,axis=0))
print("Row means:",np.mean(marks_2d,axis=1))
```

```python
import pandas as pd 

#Creating a series
marks=pd.Series([85,72,91,65,78])
print("Series :\n",marks)
print("\n Type:",type(marks))
print("Values",marks.values)
print("Index:",marks.index)
```

```python
#Series with the custom index
marks=pd.Series(
    [85,72,91,65,78],
    index=["Naveen","Akash","Devika","Shreya","Nithin"])
print("Series with name:")
print(marks)
print("\nNaveen's marks:", marks["Naveen"])
print("Marks above 75:",marks[marks>75])
```

```python
#Filtering with Pandas
df=pd.read_csv("Students.csv")

# Filter students who are passed

passed=df[df["Marks"]>=60]
print("Passed students:")
print(passed)

#Filtering students from Bemgaluru
bengaluru=df[df["City"]=="Bengaluru"]
print("\nstudents froooom Bengaluru are:")
print(bengaluru)

#Multipleconditions
passed_bengaluru=df[(df["Marks"]>=60) & (df["City"]=="Bengaluru")]
print("\nPassed students from bengaluru are:")
print(passed_bengaluru)
```

```python
import numpy as np
scores = np.array([82, 67, 91, 74, 88, 95, 69, 77, 84, 72])
print("Sum:", np.sum(scores))
print("Mean:", np.mean(scores))
print("Max:", np.max(scores))
print("Min:", np.min(scores))
above_75 = scores[scores > 75]
print("Scores above 75:", above_75)
desc_sorted = np.sort(scores)[::-1]
print("Descending order:", desc_sorted)
marks_2d = np.array([
    [85, 78, 92],
    [88, 74, 81],
    [90, 89, 84]
])
print("\n2D Array:\n", marks_2d)
col_means = np.mean(marks_2d, axis=0)
print("Subject averages:", col_means)
row_means = np.mean(marks_2d, axis=1)
print("Student averages:", row_means)
```

```python
import pandas as pd 
data={
    "Name":["Naeven","Akash","Devika","Shreya","Nithin","vidarth","vinay","Nandan"],
    "Age":[20,21,19,22,20,21,21,22],
    "Marks":[85,72,91,65,78,89,92,96],
    "City":["Bengaluru","Mysore","Bengaluru","Hubli","Mangaluru","Dharwada","Shivamogga","Shivamogga"]
    
}
df=pd.DataFrame(data)
print("Shape:",df.shape)
print("\nData types:\n",df.dtypes)
print("\nBasic statistics:\n",df.describe())
print("\nINFO")
df.info()
df["Result"]=df["Marks"].apply(lambda x: "Pass" if x>=60 else "Fail")
df["grade"]=df["Marks"].apply(lambda x: "A" if x>=80 else "B" if x>= 60 else "C")
df.to_csv("student_pandas.csv")
print("student_pandas.csv creadted")
print(df)

```
